```python
import json 
import pandas as pd
import os
import uuid
```

```python
data_list = []
base_path = '/Users/log/Github/sketchvlm/datasets/connect_dots_dataset'

# Define all the subdirectories and their types
datasets = [
    ('outlines_gt', 'outlines'),
    ('outlines_source', 'outlines'),
    ('random_gt', 'random'),
    ('random_source', 'random'),
    ('worksheets_gt', 'worksheets'),
    ('worksheets_source', 'worksheets')
]

for subdir, dataset_type in datasets:
    dataset_path = os.path.join(base_path, subdir)
    for file in os.listdir(dataset_path):
        if file.endswith('.json'):
            with open(os.path.join(dataset_path, file), 'r') as f:
                data = json.load(f)
            
            # Remove fields we don't want
            data.pop('closed_path', None)
            data.pop('prompt_txt', None)
            
            # Add standardized fields
            data['uuid'] = str(uuid.uuid4())
            data['name'] = file.replace('.json', '')
            data['image_path'] = os.path.join(dataset_path, file.replace('.json', '.png'))
            data['type'] = dataset_type
            data['variant'] = subdir
            
            data_list.append(data)

df = pd.DataFrame(data_list)
df.to_csv('connect_dots_dataset.csv', index=False)
```
